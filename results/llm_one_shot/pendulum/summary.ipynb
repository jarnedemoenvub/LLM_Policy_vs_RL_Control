```python
import pandas as pd
```

```python
# read csv file summary.csv
df = pd.read_csv('summary.csv')
```

```python
df.head()
```

```python
df_valid = df[df["n_failures"] == 0]

df_summary = df.groupby("temperature").apply(
    lambda g: pd.Series({
        "mean_reward_mean": g.loc[g["n_failures"] == 0, "mean_reward"].mean(),
        "mean_reward_std": g.loc[g["n_failures"] == 0, "mean_reward"].std(),
        "valid_runs": (g["n_failures"] == 0).sum(),
        "best_run": g.loc[g["n_failures"] == 0, "mean_reward"].max(),
    })
).reset_index()

df_summary = df_summary.round(3)
df_summary
```
